```python
# ==========================================
# CRUD API PROJECT - CLEAN FINAL VERSION
# ==========================================

!pip install fastapi sqlalchemy uvicorn -q

from fastapi import FastAPI, HTTPException
from pydantic import BaseModel, ConfigDict
from sqlalchemy import create_engine, Integer, Column, String, Boolean
from sqlalchemy.orm import declarative_base, sessionmaker


# ------------------------------------------
# 1. FastAPI App
# ------------------------------------------

app = FastAPI(title="CRUD API Project")


# ------------------------------------------
# 2. Database Setup
# ------------------------------------------

DATABASE_URL = "sqlite:///./todo.db"

engine = create_engine(
    DATABASE_URL,
    connect_args={"check_same_thread": False}
)

Base = declarative_base()

SessionLocal = sessionmaker(
    autocommit=False,
    autoflush=False,
    bind=engine
)


# ------------------------------------------
# 3. Database Model
# ------------------------------------------

class TaskDB(Base):
    __tablename__ = "tasks"

    id = Column(Integer, primary_key=True)
    title = Column(String)
    is_done = Column(Boolean, default=False)


Base.metadata.create_all(bind=engine)


# ------------------------------------------
# 4. Pydantic Model
# ------------------------------------------

class Task(BaseModel):
    id: int
    title: str
    is_done: bool = False

    model_config = ConfigDict(from_attributes=True)


# ------------------------------------------
# 5. CREATE - POST
# ------------------------------------------

@app.post("/tasks")
def create_task(task: Task):
    db = SessionLocal()

    new_task = TaskDB(
        id=task.id,
        title=task.title,
        is_done=task.is_done
    )

    db.add(new_task)
    db.commit()
    db.refresh(new_task)
    db.close()

    return new_task


# ------------------------------------------
# 6. READ - GET
# ------------------------------------------

@app.get("/tasks")
def get_tasks():
    db = SessionLocal()

    tasks = db.query(TaskDB).all()

    db.close()

    return tasks


# ------------------------------------------
# 7. UPDATE - PUT
# ------------------------------------------

@app.put("/tasks/{task_id}")
def update_task(task_id: int, task: Task):
    db = SessionLocal()

    db_task = db.query(TaskDB).filter(
        TaskDB.id == task_id
    ).first()

    if db_task is None:
        db.close()

        raise HTTPException(
            status_code=404,
            detail="Task not found"
        )

    db_task.title = task.title
    db_task.is_done = task.is_done

    db.commit()
    db.refresh(db_task)
    db.close()

    return db_task


# ------------------------------------------
# 8. DELETE - DELETE
# ------------------------------------------

@app.delete("/tasks/{task_id}")
def delete_task(task_id: int):
    db = SessionLocal()

    db_task = db.query(TaskDB).filter(
        TaskDB.id == task_id
    ).first()

    if db_task is None:
        db.close()

        raise HTTPException(
            status_code=404,
            detail="Task not found"
        )

    db.delete(db_task)
    db.commit()
    db.close()

    return {
        "message": "Task deleted successfully"
    }


# ------------------------------------------
# 9. Success Message
# ------------------------------------------

print("CRUD API Project loaded successfully!")
print("Available endpoints:")
print("POST   /tasks")
print("GET    /tasks")
print("PUT    /tasks/{task_id}")
print("DELETE /tasks/{task_id}")
```
